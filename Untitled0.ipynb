```python
messages = [
    "Hi, this is James. Can I come Friday at 3 for a haircut?",
    "Need a fade tomorrow morning. I'm Malik.",
    "Hello, do you have space Saturday?",
    "Can someone call me back?",
    "I want braids on Sunday after church. Name is Amina.",
    "Friday works, but also maybe Monday. Not sure yet.",
    "Book me in.",
    "This is Priya. Cut and color, next Thursday at 1.",
    "Are you open?",
    "My son needs a cut this weekend. I am David.",
]

days = ["monday", "tuesday", "wednesday", "thursday", "friday", "saturday", "sunday", "tomorrow", "weekend"]
services = ["haircut", "cut", "fade", "braids", "color"]

def find_name(lower):
    for marker in ["this is ", "i'm ", "i am ", "name is "]:
        if marker in lower:
            after = lower.split(marker, 1)[1]
            return after.split(".")[0].split(",")[0].strip()
    return "missing"

def check(text):
    lower = text.lower()
    found_days = [day for day in days if day in lower]
    found_services = [service for service in services if service in lower]
    name = find_name(lower)
    day = found_days[0] if len(found_days) == 1 else "missing"
    request = found_services[0] if found_services else "missing"
    flag = "OK" if name != "missing" and day != "missing" and request != "missing" else "ASK_HUMAN"
    return name, day, request, flag

for i, message in enumerate(messages, start=1):
    name, day, request, flag = check(message)
    print(i, "|", name, "|", day, "|", request, "|", flag)
```
